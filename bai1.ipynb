```python
class Employee:
    def __init__(self, name, salary, department):
        self.name = name                  
        self.__salary = salary             
        self._department = department      

    def get_salary(self):
        return self.__salary

    def increase_salary(self, amount):
        if amount > 0:
            self.__salary += amount

    def calculate_bonus(self):
        return self.__salary * 0.05

    def show_info(self):
        print(f"Tên: {self.name}")
        print(f"Lương: {self.get_salary():,.0f}")
        print(f"Phòng ban: {self._department}")
        print(f"Thưởng: {self.calculate_bonus():,.0f}")


class Developer(Employee):
    def __init__(self, name, salary, department, programming_language, overtime_hours):
        super().__init__(name, salary, department)
        self.programming_language = programming_language
        self.overtime_hours = overtime_hours

    def calculate_bonus(self):
        return self.get_salary() * 0.10 + self.overtime_hours * 100000

    def show_info(self):
        super().show_info()
        print(f"Ngôn ngữ lập trình: {self.programming_language}")
        print(f"Số giờ làm thêm: {self.overtime_hours}")
        print()


class Manager(Employee):
    def __init__(self, name, salary, department, number_of_employees):
        super().__init__(name, salary, department)
        self.number_of_employees = number_of_employees

    def calculate_bonus(self):
        return self.get_salary() * 0.15 + self.number_of_employees * 200000

    def show_info(self):
        super().show_info()
        print(f"Số nhân viên quản lý: {self.number_of_employees}")
        print()



employees = [
    Developer("An", 15000000, "IT", "Python", 10),
    Developer("Bình", 18000000, "IT", "C++", 5),
    Manager("Cường", 25000000, "Kỹ thuật", 10),
    Manager("Dũng", 22000000, "Kinh doanh", 8)
]

print("===== THÔNG TIN NHÂN VIÊN =====")

for employee in employees:
    employee.show_info()

highest_salary_employee = employees[0]

for employee in employees:
    if employee.get_salary() > highest_salary_employee.get_salary():
        highest_salary_employee = employee

print("===== NHÂN VIÊN CÓ LƯƠNG CAO NHẤT =====")
print(highest_salary_employee.name)
print(f"Lương: {highest_salary_employee.get_salary():,.0f}")

total_bonus = 0

for employee in employees:
    total_bonus += employee.calculate_bonus()

print("\n===== TỔNG TIỀN THƯỞNG =====")
print(f"{total_bonus:,.0f}")

developer_count = 0
manager_count = 0

for employee in employees:
    if isinstance(employee, Developer):
        developer_count += 1

    if isinstance(employee, Manager):
        manager_count += 1

print("\n===== SỐ LƯỢNG =====")
print("Developer:", developer_count)
print("Manager:", manager_count)
```
