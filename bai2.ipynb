```python
class Character:
    def __init__(self, name, hp, level):
        self.name = name              
        self.__hp = hp              
        self._level = level           

    def get_hp(self):
        return self.__hp

    def take_damage(self, damage):
        if damage > 0:
            self.__hp -= damage

            if self.__hp < 0:
                self.__hp = 0

    def heal(self, amount):
        if amount > 0:
            self.__hp += amount

    def attack(self):
        return 0

    def show_info(self):
        print(f"Tên: {self.name}")
        print(f"HP: {self.get_hp()}")
        print(f"Level: {self._level}")


class Warrior(Character):
    def __init__(self, name, hp, level, strength):
        super().__init__(name, hp, level)
        self.strength = strength

    def attack(self):
        damage = self._level * 5 + self.strength
        return damage

    def show_info(self):
        super().show_info()
        print(f"Strength: {self.strength}")
        print()


class Mage(Character):
    def __init__(self, name, hp, level, mana, magic_power):
        super().__init__(name, hp, level)
        self.__mana = mana
        self.magic_power = magic_power

    def get_mana(self):
        return self.__mana

    def attack(self):
        if self.__mana >= 10:
            self.__mana -= 10
            damage = self._level * 3 + self.magic_power
            return damage
        else:
            return 0

    def show_info(self):
        super().show_info()
        print(f"Mana: {self.get_mana()}")
        print(f"Magic Power: {self.magic_power}")
        print()



warrior1 = Warrior("Warrior A", 150, 10, 20)
warrior2 = Warrior("Warrior B", 180, 8, 25)

mage1 = Mage("Mage A", 120, 12, 50, 30)
mage2 = Mage("Mage B", 140, 9, 40, 35)

characters = [warrior1, warrior2, mage1, mage2]


print("===== THÔNG TIN BAN ĐẦU =====")

for character in characters:
    character.show_info()


print("===== CHIẾN ĐẤU =====")

damage = warrior1.attack()
mage1.take_damage(damage)
print(f"{warrior1.name} gây {damage} damage cho {mage1.name}")

damage = mage1.attack()
warrior1.take_damage(damage)
print(f"{mage1.name} gây {damage} damage cho {warrior1.name}")

damage = warrior2.attack()
mage2.take_damage(damage)
print(f"{warrior2.name} gây {damage} damage cho {mage2.name}")

damage = mage2.attack()
warrior2.take_damage(damage)
print(f"{mage2.name} gây {damage} damage cho {warrior2.name}")


print("\n===== SAU CHIẾN ĐẤU =====")

for character in characters:
    character.show_info()


strongest_hp = characters[0]

for character in characters:
    if character.get_hp() > strongest_hp.get_hp():
        strongest_hp = character

print("===== NHÂN VẬT CÒN NHIỀU HP NHẤT =====")
print(strongest_hp.name)
print("HP:", strongest_hp.get_hp())


print("\n===== KIỂM TRA isinstance() =====")

for character in characters:
    if isinstance(character, Warrior):
        print(character.name, "là Warrior")

    if isinstance(character, Mage):
        print(character.name, "là Mage")


print("\n===== KIỂM TRA issubclass() =====")

print("Warrior kế thừa Character:",
      issubclass(Warrior, Character))

print("Mage kế thừa Character:",
      issubclass(Mage, Character))

print("Character kế thừa Warrior:",
      issubclass(Character, Warrior))
```
